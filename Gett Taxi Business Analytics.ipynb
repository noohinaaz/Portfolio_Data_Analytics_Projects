<strong style="font-size: 22px; color: #253338;">🧩 Bridging Business Failures with Microservices Orchestration Logic</strong><br><br>

The Gett Taxi platform operates within a distributed microservices architecture where core processes like matching, offer dispatch, and ride assignment rely heavily on seamless inter-service communication. The central orchestration system—tasked with coordinating these services—becomes critical in ensuring that an order transitions smoothly from creation to fulfillment.<br><br>

In this context, the failures observed in our dataset—such as driver assignment issues, high cancellation rates, and ETA delays—are not just operational hiccups. They signal breakdowns or bottlenecks in the orchestration layer, where one service's delay or non-response disrupts the entire order lifecycle. Especially during peak hours, the orchestrator may struggle to maintain performance, triggering compensatory flows like order rejections or delayed cancellations.<br><br>

Thus, our business analysis doesn’t just expose service-level patterns—it maps directly to orchestration challenges such as:<br>
• <strong>Centralized failure handling through compensation</strong> (e.g., automatic cancellation)<br>
• <strong>Service dependency latency</strong> (e.g., delayed driver assignment)<br>
• <strong>Lack of observability under high load</strong> (e.g., inconsistent ETA responses)

Relevence to the project:

• <strong>High cancellation before driver assignment may stem from orchestration timeouts in the MatchingService.</strong><br>
• <strong>Prolonged ETA and cancellation with driver suggest slow orchestration or compensatory rollback delays.</strong><br>
• <strong>Peak-hour instability could stress the orchestrator, causing slower responses and increased failure rates.</strong><br>

```python
df["is_driver_assigned"] = np.where(df["is_driver_assigned_key"] == 1, "Yes", "No")
df["order_status"] = np.where(df["order_status_key"] == 4, "Client Cancelled", "System Reject")

df.drop(columns=["is_driver_assigned_key", "order_status_key"], inplace=True)
```

```python
unique_coords = df[['origin_latitude', 'origin_longitude']].drop_duplicates().reset_index(drop=True)
coord_tuples = list(zip(unique_coords['origin_latitude'], unique_coords['origin_longitude']))
results = rg.search(coord_tuples)  # This is fast and efficient
unique_coords['location'] = [res['name'] for res in results]
df = df.merge(unique_coords, on=['origin_latitude', 'origin_longitude'], how='left')
```

```python
df['location'] = df['location'].str.replace(r',?\s*England$', '', regex=True)
```

```python
location_to_region = {
    'Reading': 'Central',
    'Earley': 'East',
    'Lower Earley': 'East',
    'Shinfield': 'South',
    'Tilehurst': 'West',
    'Calcot': 'West',
    'Woodley': 'East',
    'Sonning': 'North East',
    'Sonning Common': 'North',
    'Winnersh': 'South East',
    'Swallowfield': 'South',
    'Theale': 'Far West',
    'Barkham': 'South West'}
df['region'] = df['location'].map(location_to_region)
```

```python
df["order_hour"] = df["order_datetime"].str.split(":").apply(lambda split: split[0]).astype("int64")
```

```python
df.loc[df["is_driver_assigned"] == "No", ["m_order_eta", "cancellations_time_in_seconds"]] = \
    df.loc[df["is_driver_assigned"] == "No", ["m_order_eta", "cancellations_time_in_seconds"]].fillna(0)
```

```python
df['order_eta_status'] = df['m_order_eta'].apply(lambda x: "Late" if x > 240 else "Under Mean Time")
```

```python
df["cancellations_time_in_seconds"].describe(percentiles=[0,0.75, 0.80, 0.85, 0.90, 0.95, 0.99])
```

```python
threshold = df["cancellations_time_in_seconds"].quantile(0.99)
df = df[df["cancellations_time_in_seconds"] <= threshold]
df.shape
```

```python
threshold = df["m_order_eta"].quantile(0.97)
df = df[df["m_order_eta"] <= threshold]
df.shape
```

```python
bins = list(range(0, 901, 60))  
labels = list(range(1, 16)) 
df["minutes"] = pd.cut(df["cancellations_time_in_seconds"], bins=bins, labels=labels, right=True)
```
